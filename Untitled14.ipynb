```python
import pandas as pd
import matplotlib.pyplot as plt

# Student data
data = {
    "Student": ["Sara", "Ahmed", "Nora", "Ali", "Lama"],
    "Grade": [95, 80, 88, 75, 92]
}

df = pd.DataFrame(data)

# Analyze grades
print(df)
print("Average:", df["Grade"].mean())
print("Highest Grade:", df["Grade"].max())
print("Lowest Grade:", df["Grade"].min())

# Visualize grades
plt.bar(df["Student"], df["Grade"], color="skyblue")
plt.title("Student Grade Analysis")
plt.xlabel("Students")
plt.ylabel("Grades")
plt.ylim(0, 100)
plt.show()
```
